```python
# CELL 1 — Hubungkan notebook ke repository publik package
# Ganti dengan owner/repository GitHub yang sudah kamu publikasikan.
GITHUB_REPO = "https://github.com/Lipzz123/qwen-image-edit-project"
BRANCH = "main"

import os, subprocess, sys
from pathlib import Path

if GITHUB_REPO.startswith("YOUR_") or "/" not in GITHUB_REPO:
    raise ValueError("Edit GITHUB_REPO terlebih dahulu, contoh: username/qwen-image-edit-project")

REPO_DIR = Path("/content/qwen-image-edit-project")
if REPO_DIR.exists():
    subprocess.run(["git", "-C", str(REPO_DIR), "fetch", "origin", BRANCH], check=True)
    subprocess.run(["git", "-C", str(REPO_DIR), "reset", "--hard", f"origin/{BRANCH}"], check=True)
else:
    subprocess.run(["git", "clone", "--depth", "1", "--branch", BRANCH, f"https://github.com/{GITHUB_REPO}.git", str(REPO_DIR)], check=True)

print("Repository siap:", REPO_DIR)
```

## Troubleshooting
- **GPU tidak terdeteksi:** pilih GPU di Runtime → Change runtime type, restart runtime, lalu Run all.
- **Disk tidak cukup:** runtime Colab harus memiliki ruang kosong yang cukup untuk model, cache, dan file sementara.
- **Node tidak ditemukan:** lihat output instalasi ComfyUI-GGUF; restart runtime lalu jalankan ulang.
- **Generate CUDA out of memory:** coba resolusi 768, tutup proses lain, restart runtime, lalu ulangi setup.
- **Model download gagal:** periksa koneksi internet, ketersediaan repository Hugging Face, nama file, dan ketentuan akses model.
- **Hasil tidak muncul:** periksa `/content/qwen_image_2_1/comfyui.log`. Hasil tersimpan di `/content/qwen_image_2_1/outputs`.

Notebook dan package ini merupakan template implementasi. Eksekusi penuh harus diuji pada runtime Colab aktual karena versi ComfyUI, custom node, dependency, GPU, dan ketersediaan file model dapat berubah.
